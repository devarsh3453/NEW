```python
import random
```

```python
from abc import ABC, abstractmethod # abc = abstract base class
```

```python
class Runnable(ABC):

  @abstractmethod
  def invoke(input_data):
        pass
```

```python
class My_llm(Runnable):
  def __init__(self):
    print('LLM created')

  def invoke(self,input_data):
    return self.predict(input_data)

  def predict(self,prompt):
    resp_list=[
        'A',
        'B',
        'C'
    ]
    result= "AI: " +prompt
    return {'response':  result}
```

```python
llm=My_llm()
```

```python
class My_prompt(Runnable):
  def __init__(self,template,input):
    self.template=template
    self.input=input


  def invoke(self,input_data):
    return self.formate(input_data)

  def formate(self,input_dict):
    return self.template.format(**input_dict)
```

```python
class My_parser(Runnable):
  def __init__(self):
    pass

  def invoke(self,input_data):
    return input_data['response']

```

#So now we have standerdise our components and can add any number of components


```python
class RunnableConnector(Runnable):

  def __init__(self,runnable_list):
    self.runnable_list=runnable_list

  def invoke(self,input_data):

    for runnable in self.runnable_list:
      input_data=runnable.invoke(input_data)
    return input_data
```

```python
template=My_prompt(
    template="Give me details of {topic} in {n} Paras",
    input=['n','topic']
)
```

```python
llm=My_llm()
```

```python
parser=My_parser()
```

```python
chain=RunnableConnector([template,llm])
chain.invoke({'topic':'python','n':12})
```

```python
chain=RunnableConnector([template,llm,parser])
chain.invoke({'topic':'python','n':12})
```
