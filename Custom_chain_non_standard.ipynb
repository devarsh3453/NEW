```python
import random
```

```python
class My_llm:
  def __init__(self):
    print('LLM created')

  def predict(self,prompt):
    resp_list=[
        'A',
        'B',
        'C'
    ]
    return {'response': random.choice(resp_list)}
```

```python
llm=My_llm()
llm.predict('x')
```

```python
class My_prompt:
  def __init__(self,template,input):
    self.template=template
    self.input=input

  def formate(self,input_dict):
    return self.template.format(**input_dict)
```

```python
template=My_prompt(
    template="give me {n} points on {topic}",
    input=['n','topic']
)
```

```python
prompt=template.formate({'n':10,'topic':'max'})
```

```python
llm.predict(prompt)
```

# But will need different chains for each component differentely


```python
class My_chain:
  def __init__(self,llm,prompt):
    self.llm=llm
    self.prompt=prompt

  def run(self,input_dict):
    prompt=self.prompt.formate(input_dict)
    result=self.llm.predict(prompt)
    return result['response']


```

```python
template1=My_prompt(
    template="give me {n} points on {topic}",
    input=['n','topic']
)
llm=My_llm()
```

```python
chain1=My_chain(llm=llm,prompt=template1)
```

```python
chain1.run({'n':12,'topic':'india'})
```
